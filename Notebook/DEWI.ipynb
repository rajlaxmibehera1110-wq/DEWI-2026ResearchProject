```python
# ==========================================================
#              DATA CLEANING - DEWI PROJECT
# ==========================================================

# -----------------------------
# Import Required Libraries
# -----------------------------
import pandas as pd
import numpy as np

# -----------------------------
# Load Dataset 1
# -----------------------------
df1 = pd.read_csv("Crime against Women (2001-2022).csv")

# -----------------------------
# Load Dataset 2
# -----------------------------
df2 = pd.read_csv("CrimesOnWomenData.csv")

# -----------------------------
# Remove Unnecessary Columns
# -----------------------------
if 'Unnamed: 0' in df2.columns:
    df2.drop(columns=['Unnamed: 0'], inplace=True)

# -----------------------------
# Display Dataset Information
# -----------------------------
print("Dataset 1")
print(df1.info())

print("\nDataset 2")
print(df2.info())

# -----------------------------
# Check Missing Values
# -----------------------------
print("\nMissing Values - Dataset 1")
print(df1.isnull().sum())

print("\nMissing Values - Dataset 2")
print(df2.isnull().sum())

# -----------------------------
# Check Duplicate Records
# -----------------------------
print("\nDuplicate Rows - Dataset 1:", df1.duplicated().sum())
print("Duplicate Rows - Dataset 2:", df2.duplicated().sum())

# -----------------------------
# Remove Duplicate Records
# -----------------------------
df1 = df1.drop_duplicates()
df2 = df2.drop_duplicates()

# -----------------------------
# Remove Extra Spaces
# -----------------------------
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

for col in df1.select_dtypes(include='object'):
    df1[col] = df1[col].str.strip()

for col in df2.select_dtypes(include='object'):
    df2[col] = df2[col].str.strip()

# -----------------------------
# Check Data Types
# -----------------------------
print("\nDataset 1 Data Types")
print(df1.dtypes)

print("\nDataset 2 Data Types")
print(df2.dtypes)

# -----------------------------
# Statistical Summary
# -----------------------------
print("\nDataset 1 Summary")
print(df1.describe())

print("\nDataset 2 Summary")
print(df2.describe())

# -----------------------------
# Final Shape After Cleaning
# -----------------------------
print("\nDataset 1 Shape:", df1.shape)
print("Dataset 2 Shape:", df2.shape)

print("\nData Cleaning Completed Successfully!")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import LabelEncoder
import numpy as np

# --- Prepare Dataset 1 ---
# Melt the DataFrame to a long format suitable for regression
df1_melted = df1_original.melt(id_vars=['CRIME HEAD'], var_name='Year', value_name='Crime_Count')

# Convert 'Year' to numeric type
df1_melted['Year'] = pd.to_numeric(df1_melted['Year'])

# Encode 'CRIME HEAD'
le_crime_head = LabelEncoder()
df1_melted['CRIME_HEAD_Encoded'] = le_crime_head.fit_transform(df1_melted['CRIME HEAD'])

# Define training and testing years
train_years_df1_numeric = range(2001, 2019)
test_years_df1_numeric = range(2019, 2023)

# Split into training and testing dataframes
df1_train_prepared = df1_melted[df1_melted['Year'].isin(train_years_df1_numeric)]
df1_test_prepared = df1_melted[df1_melted['Year'].isin(test_years_df1_numeric)]

# Define features (X) and target (y)
X_train_df1 = df1_train_prepared[['CRIME_HEAD_Encoded', 'Year']]
y_train_df1 = df1_train_prepared['Crime_Count']
X_test_df1 = df1_test_prepared[['CRIME_HEAD_Encoded', 'Year']]
y_test_df1 = df1_test_prepared['Crime_Count']

# Initialize and train the Linear Regression model
model_df1 = LinearRegression()
model_df1.fit(X_train_df1, y_train_df1)

# Make predictions on the test set
y_pred_df1 = model_df1.predict(X_test_df1)

# Evaluate the model
mae_df1 = mean_absolute_error(y_test_df1, y_pred_df1)
mse_df1 = mean_squared_error(y_test_df1, y_pred_df1)
r2_df1 = r2_score(y_test_df1, y_pred_df1)

print("\n--- Dataset 1 Linear Regression Model Evaluation ---")
print(f"Mean Absolute Error (MAE): {mae_df1:.2f}")
print(f"Mean Squared Error (MSE): {mse_df1:.2f}")
print(f"R-squared (R2): {r2_df1:.2f}")
```
